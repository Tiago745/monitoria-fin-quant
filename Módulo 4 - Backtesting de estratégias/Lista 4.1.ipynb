## **4.1 - Introdução ao Backtesting**

### **O que é backtest e por que ele é essencial?**

Imagem faltando.jpg

Backtest é o processo de aplicar um conjunto de regras de negociação (uma estratégia) a dados históricos para simular como essa estratégia teria se comportado no passado. É a ferramenta mais importante no arsenal de um pesquisador quantitativo, pois permite testar uma ideia de investimento em um ambiente de laboratório, sem arriscar capital real.

Uma boa analogia para pensarmos o backtest é comparar o processo de pesquisa em um fundo quant a uma pesquisa científica, o backtest seria o teste de hipótese, talvez a parte mais importante de toda a pesquisa. Um erro ou imprecisão nessa fase pode levar à perda de milhões para os responsáveis, então robustez é essencial agora.

Apesar de parecer relativamente, precisamos apenas comparar preços de compra e venda e calcular os retornos, há muitos viéses que devem ser observados e tidos em consideração na confecção do backtest e é isso que estudaremos aqui.

Segue um breve resumo dos principais pontos a serem abordados!

1. Dados Históricos Confiáveis
A qualidade do backtest é diretamente proporcional à qualidade dos dados. Dados ruins ou incompletos levarão a conclusões falsas. Os dados precisam ser:
* **Precisos e Limpos:** Sem falhas, valores ausentes ou erros.
* **Ajustados:** Devem considerar eventos corporativos como **dividendos, desdobramentos (splits) e grupamentos (inplits)**, que alteram o preço de um ativo sem que haja uma negociação.
* **Abrangentes:** Cobrir um longo período de tempo que inclua diferentes regimes de mercado (crises, euforia, períodos de lado, etc.).
* **Granularidade Adequada:** A frequência dos dados (diária, por hora, por minuto ou até por tick) deve ser apropriada para a estratégia que está sendo testada.

 2. O Motor de Simulação (Backtesting Engine)
Ele é o "cérebro"  do sistema. É um software ou código que executa as seguintes tarefas:
* **Itera sobre os Dados:** Ele avança no tempo, barra por barra de preço (ex: dia a dia).
* **Gera Sinais:** A cada passo, ele alimenta a lógica da estratégia com os dados disponíveis até aquele momento.
* **Executa Ordens:** Quando a estratégia gera um sinal de "comprar" ou "vender", o motor simula a execução dessa ordem, considerando premissas como custos e slippage (derrapagem).
* **Gerencia o Portfólio:** Ele mantém o controle de uma carteira virtual, calculando posições abertas, caixa disponível, lucros e prejuízos não realizados.
* **Calcula as Métricas:** Ao final da simulação, ele compila todos os dados e gera um relatório completo de desempenho.

 3. A Lógica da Estratégia
Este é o conjunto de regras objetivas e inequívocas que definem quando entrar ou sair de uma posição. Não há espaço para subjetividade. Um exemplo de lógica simples seria:
* **Sinal de Compra:** "Comprar 100 ações da PETR4 quando sua média móvel de 50 dias cruzar para cima da média móvel de 200 dias."
* **Sinal de Venda:** "Vender a posição quando a média móvel de 50 dias cruzar para baixo da de 200 dias."

 Os Objetivos e as Análises de um Backtest

O propósito fundamental de um backtest é traduzir uma ideia de trading em números concretos, permitindo uma análise multifacetada do seu potencial. As perguntas que um backtest responde são agrupadas em três grandes áreas de análise:

 Análise de Rentabilidade: A Estratégia Gera Lucro? 

Esta é a primeira verificação. Se a estratégia não gera lucro no papel, não há motivo para continuar.
* **Retorno Total e Anualizado (CAGR):** Mostra o crescimento percentual total do capital e qual seria a taxa de crescimento anual composta. É a principal medida de desempenho.
* **Fator de Lucro (Profit Factor):** Calculado como `Lucro Bruto / Prejuízo Bruto`. Um valor de 2.0 significa que a estratégia gerou R$ 2,00 de lucro para cada R$ 1,00 de prejuízo. Valores consistentemente acima de 1.5 são geralmente considerados bons.
* **Expectativa Matemática por Operação:** Indica o valor financeiro médio que você pode esperar ganhar (ou perder) em cada operação. Uma expectativa positiva é crucial para a viabilidade a longo prazo.

 Análise de Risco: Qual o Custo do Retorno? 

Uma alta rentabilidade com um risco gigantesco pode não ser desejável. Esta análise mede a "qualidade" do retorno.
* **Drawdown Máximo:** Esta é talvez a métrica de risco mais importante. Ela mede a **maior perda percentual do pico ao fundo** que a carteira sofreu durante a simulação. Responde à pergunta: "Qual foi a pior dor financeira que eu teria suportado para obter esse retorno?".
* **Duração do Drawdown:** Mede quanto tempo a estratégia levou para se recuperar e atingir um novo pico de capital após o início de um drawdown. Períodos longos podem ser psicologicamente difíceis de suportar.
* **Volatilidade dos Retornos:** Mede o quão "turbulento" foi o caminho para o lucro. Alta volatilidade significa grandes oscilações no valor da carteira.
* **Índices de Risco-Retorno:**
    * **Índice de Sharpe:** Mede o retorno excedente (acima de um ativo livre de risco) por unidade de volatilidade total. É o mais conhecido.
    * **Índice de Sortino:** Similar ao Sharpe, mas foca apenas na volatilidade negativa (dos retornos ruins), pois a maioria dos investidores não se importa com a "volatilidade para cima".
    * **Information Ratio:** Algo similar ao Sharpe mas considerando apenas o alpha e não o retorno completo.

 Análise de Consistência e Comportamento: O Desempenho é Robusto? 

Esta análise investiga como a estratégia se comporta e se seu desempenho é consistente ao longo do tempo.
* **Taxa de Acerto (Win Rate):** A porcentagem de operações que terminaram com lucro. É importante notar que uma taxa de acerto baixa (ex: 40%) pode ser muito lucrativa se os ganhos médios forem muito maiores que as perdas médias.
* **Relação Média Ganhos/Perdas (Average Win/Loss Ratio):** Mostra o tamanho do seu ganho médio em relação à sua perda média. Uma relação de 3:1 indica que, em média, seus trades vencedores são três vezes maiores que os perdedores.
* **Número de Operações:** Informa a frequência da estratégia. Uma estratégia com pouquíssimas operações pode ter sua validade estatística questionada, enquanto uma com muitas operações pode ter seus lucros consumidos pelos custos de transação.
* **Análise por Período:** Permite avaliar como a estratégia se saiu em diferentes cenários de mercado (ex: Crise de 2008, pandemia de 2020, mercados em alta). Uma estratégia robusta deve ser capaz de, no mínimo, sobreviver a diferentes condições.

### **O que é um retorno?**

Apesar de ser intuitivo o que é um retorno, é de suma importância que isso seja definido formalmente para que possamos explorar as propriedades matemáticas dos retornos e para que utilizemos os modelos de forma correta, a simbologia matemática foi utilizada para ser o mais didático possível, mas caso tenha dúvidas não exite em chamar alguém da gestão!

Denotamos o valor do ativo amanhã (ou em algum período futuro) como $V_0(1 + r_i)$. Uma maneira equivalente de definir os retornos é a partir dos preços de fechamento do título $i$ nos dias $0$ e $1$, que denotamos como $P_{i,0}$ e $P_{i,1}$, respectivamente. O retorno é definido como:

$$r_{i,1} := \frac{P_{i,1} - P_{i,0}}{P_{i,0}}$$

Estendemos essa definição para o caso em que o título paga um dividendo. O detentor do ativo recebe um valor $D_{i,1}$. O retorno ajustado por dividendos é definido como:

$$r_{i,1} := \frac{P_{i,1} + D_{i,1} - P_{i,0}}{P_{i,0}}$$

Em um universo de $n$ ativos, o vetor de retornos diários entre os tempos $t-1$ e $t$ é denotado por $\mathbf{r}_t := (r_{1,t}, \dots, r_{n,t})$. Para um portfólio $\mathbf{w} \in \mathbb{R}^n$, onde $w_i$ é um valor monetário investido no ativo $i$, o PnL (Lucros e Perdas) em um único período é dado pela variação no valor do portfólio. O número de ações detidas no ativo $i$ é calculado como $w_i / P_{i,0}$. O valor do portfólio no período um é $\sum_i \left(\frac{w_i}{P_{i,0}}\right) P_{i,1}$, e a variação no valor é:

$$\sum_i w_i \frac{P_{i,1}}{P_{i,0}} - \sum_i w_i$$

Em forma de vetor, isso é igual a $\mathbf{w}^\mathsf{T}\mathbf{r}$.

### Volatilidade e Retornos Compostos

A variância é o quadrado da volatilidade. Ocasionalmente, quando a aproximação $E[r^2] \gg (E[r])^2$ se sustenta, aproximaremos a volatilidade pelo segundo momento do retorno: $\sigma^2 \approx E[r^2]$.

O retorno composto é definido como o valor de uma unidade de numerário após um período $T$, se tivesse sido investido em um título com retornos $r_t$, ou seja:

$$
\begin{aligned}
R_{1:T} &:= \frac{P_T}{P_0} - 1 \\
&= \frac{P_T}{P_{T-1}} \times \frac{P_{T-1}}{P_{T-2}} \times \dots \times \frac{P_1}{P_0} - 1 \\
&= \prod_{t=1}^{T} (1 + r_t) - 1
\end{aligned}
$$


### Retornos Brutos vs. Log-Retornos na Prática

Se $\mathbf{r}$ segue uma distribuição normal multivariada, então o mesmo acontece com o PnL do portfólio. A variância deste portfólio pode ser calculada usando apenas duas informações: os pesos do portfólio e a matriz de covariância dos retornos.

A questão de saber se os retornos líquidos são normais é uma questão empírica. Sabemos, pelo menos, que se os retornos líquidos são normais, eles são muito tratáveis para análise em um determinado ponto no tempo. No entanto, eles não são facilmente tratáveis na análise de séries temporais. Por exemplo, considere o retorno composto sobre os períodos $1, \dots, T$. Se os $r_{i,t}$ são normalmente distribuídos, o retorno total acumulado não é normalmente distribuído, e sua distribuição diverge rapidamente da distribuição normal. A variância dos retornos acumulados não é uma função simples das variâncias de um único período.

Por outro lado, os log-retornos são aditivos ao longo do tempo. Seja $\tilde{r}_{i,t} := \log(1 + r_{i,t})$. Então, o logaritmo do retorno composto é igual à soma dos log-retornos no mesmo período, e se o log-retorno é normal, o mesmo acontece com o logaritmo dos retornos compostos. Se os retornos são independentes, a variância dos log-retornos é igual à soma das variâncias.

Podemos reconciliar as duas visões de retornos — brutos e logarítmicos — se a aproximação $\log(x) = (x-1) + O(|x-1|^2)$ for suficientemente precisa, ou seja, se os retornos líquidos forem pequenos. Neste caso, podemos fazer a aproximação $\tilde{r}_i \approx r_i$, que é suficientemente precisa desde que os retornos não sejam muito grandes.

Uma aproximação comum para o retorno líquido composto de um ativo ao longo do tempo é dada por:
$$
\begin{aligned}
\prod_t (1 + r_t) - 1 &= \exp\left(\sum_t \tilde{r}_t\right) - 1 \\
&\approx \left(1 + \sum_t \tilde{r}_t\right) - 1 \\
&= \sum_t \tilde{r}_t
\end{aligned}
$$
Sempre verifique a precisão da aproximação, por exemplo, comparando a estimativa de modelos desenvolvidos usando $r$ e $\tilde{r}$.

### Fatos Estilizados dos Retornos (Stylized Facts)

Ao analisar empiricamente as séries de retornos, certos padrões estatísticos aparecem de forma consistente em diferentes ativos e mercados. Esses padrões, conhecidos como "fatos estilizados", são cruciais porque qualquer modelo quantitativo realista deve ser capaz de capturá-los.

1.  **"Caudas Pesadas" (Fat Tails):** A distribuição dos retornos não é normal. Eventos extremos, tanto positivos quanto negativos, ocorrem com uma frequência muito maior do que uma distribuição Gaussiana preveria. Ignorar isso leva a uma subestimação perigosa do risco de mercado.

2.  **Agrupamento de Volatilidade (Volatility Clustering):** A volatilidade não é constante. Períodos de grandes oscilações de preço (alta volatilidade) tendem a ser seguidos por mais dias de alta volatilidade. Da mesma forma, períodos de calmaria (baixa volatilidade) tendem a ser seguidos por mais dias de calmaria.

3.  **Ausência de Autocorrelação nos Retornos Brutos:** É muito difícil prever o retorno de amanhã usando apenas o retorno de hoje. A correlação entre os retornos diários é, em geral, muito próxima de zero.

4.  **Assimetria (Leverage Effect):** A volatilidade tende a aumentar mais após uma grande queda nos preços do que após uma alta de mesma magnitude. Em outras palavras, a correlação entre os retornos e as mudanças na volatilidade tende a ser negativa.

*Exercício* Desigualdade: Retorno Geométrico vs. Aritmético

Para comprovarmos a robustez da utilização dos log-retornos em comparação aos retornos tradicionais em curtos espaços de tempo(diário, semanal), a prova da desigualdade abaixo traz a intuição necessária do porquê.

$$\left[ \prod_{t=1}^{T} (1 + r_t) \right]^{\frac{1}{T}} - 1 \le \frac{1}{T} \sum_{t=1}^{T} r_t$$
